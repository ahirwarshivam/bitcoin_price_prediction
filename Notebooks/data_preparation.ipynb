```python
import requests
import pandas as pd
from datetime import datetime, timezone

url = "https://api.binance.com/api/v3/klines"

params = {
    "symbol": "BTCUSDT",
    "interval": "1d",
    "startTime": int(datetime(2026, 6, 17, tzinfo=timezone.utc).timestamp() * 1000),
    "endTime": int(datetime(2026, 8, 21, tzinfo=timezone.utc).timestamp() * 1000),
    "limit": 1000
}

response = requests.get(url, params=params)
data = response.json()

columns = [
    "Open time",
    "Open",
    "High",
    "Low",
    "Close",
    "Volume",
    "Close time",
    "Quote asset volume",
    "Number of trades",
    "Taker buy base asset volume",
    "Taker buy quote asset volume",
    "Ignore"
]

new_data = pd.DataFrame(data, columns=columns)

new_data.to_csv("btc_new_data_june_aug_2026.csv", index=False)

print(new_data.head())
print(new_data.tail())
print("Rows:", len(new_data))
```

```python
# Remove 21 August 2026
new_data["Open time"] = pd.to_datetime(new_data["Open time"], unit="ms", utc=True)

new_data = new_data[new_data["Open time"] <= "2026-08-20"]

print("Rows:", len(new_data))
print("First date:", new_data["Open time"].min())
print("Last date:", new_data["Open time"].max())
```

```python
old_data = pd.read_csv("btc_1d_data_2018_to_2025.csv")

print("Old data rows:", len(old_data))
print(old_data.head())
print(old_data.tail())
```

```python
old_data["Open time"] = pd.to_datetime(
    old_data["Open time"],
    utc=True
)

# New data ka Close time bhi datetime mein convert karna
new_data["Close time"] = pd.to_datetime(
    new_data["Close time"],
    unit="ms",
    utc=True
)
```

```python
old_data["Open time"] = pd.to_datetime(old_data["Open time"], utc=True)
new_data["Open time"] = pd.to_datetime(new_data["Open time"], utc=True)

print(old_data["Open time"].tail())
print(new_data["Open time"].head())
```

```python
final_data = pd.concat(
    [old_data, new_data],
    ignore_index=True
)

# Sort chronologically
final_data = final_data.sort_values("Open time").reset_index(drop=True)

# Remove accidental duplicate dates
final_data = final_data.drop_duplicates(
    subset=["Open time"],
    keep="first"
).reset_index(drop=True)

print("Final rows:", len(final_data))
print("First date:", final_data["Open time"].min())
print("Last date:", final_data["Open time"].max())
```

```python
final_data.to_csv(
    "btc_1d_data_2018_to_2026_aug20.csv",
    index=False
)

print("Final dataset saved successfully!")
```

```python
# Check for missing daily dates

dates = final_data["Open time"].sort_values().dt.normalize()

expected_dates = pd.date_range(
    start=dates.min(),
    end=dates.max(),
    freq="D",
    tz="UTC"
)

missing_dates = expected_dates.difference(dates)

print("Expected days:", len(expected_dates))
print("Actual rows:", len(final_data))
print("Missing dates:", len(missing_dates))

if len(missing_dates) == 0:
    print(" No missing dates!")
else:
    print(" Missing dates:")
    print(missing_dates)
```
