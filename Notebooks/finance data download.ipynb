```python
import pandas as pd

# Read the CSV
dxy = pd.read_csv("DXY.csv", header=[0, 1])

# Remove the extra first row containing "Date"
dxy = dxy.iloc[1:].copy()

# Create clean dataframe using Date + Close
dxy_clean = pd.DataFrame({
    'Date': pd.to_datetime(dxy[('Price', 'Ticker')]),
    'DXY': pd.to_numeric(dxy[('Close', 'DX-Y.NYB')], errors='coerce')
})

# Remove missing values
dxy_clean = dxy_clean.dropna().reset_index(drop=True)

print(dxy_clean.head())
print(dxy_clean.tail())
print(dxy_clean.shape)
```

```python
dxy_clean.to_csv("DXY_clean.csv")
```

```python
dxy = pd.read_csv("DXY.csv", header=[0, 1])

dxy_clean = pd.DataFrame({
    'Date': pd.to_datetime(dxy[('Price', 'Ticker')], errors='coerce'),
    'DXY': pd.to_numeric(dxy[('Close', 'DX-Y.NYB')], errors='coerce')
})

dxy_clean = dxy_clean.dropna().reset_index(drop=True)

print(dxy_clean.head())
print(dxy_clean.shape)
print(dxy_clean['Date'].min())
print(dxy_clean['Date'].max())
```

```python
vix = yf.download('^VIX', start='2018-01-01', end='2026-08-22', auto_adjust=False)
sp500 = yf.download('^GSPC', start='2018-01-01', end='2026-08-22', auto_adjust=False)
gold = yf.download('GC=F', start='2018-01-01', end='2026-08-22', auto_adjust=False)
```

```python
print("VIX:", vix.shape)
print("S&P500:", sp500.shape)
print("Gold:", gold.shape)

display(vix.head())
display(sp500.head())
display(gold.head())
```

```python
print(vix.index.min(), vix.index.max())
print(sp500.index.min(), sp500.index.max())
print(gold.index.min(), gold.index.max())
```

```python
# clean the data
```

```python
vix = vix.reset_index()
sp500 = sp500.reset_index()
gold = gold.reset_index()

vix = vix[['Date','Close']].rename(columns={'Close':'VIX'})
sp500 = sp500[['Date','Close']].rename(columns={'Close':'SP500'})
gold = gold[['Date','Close']].rename(columns={'Close':'Gold'})

for x in [vix, sp500, gold]:
    x['Date'] = pd.to_datetime(x['Date']).dt.tz_localize(None)

print(vix.head())
print(sp500.head())
print(gold.head())
```

```python
vix.to_csv('VIX.csv', index=False)
sp500.to_csv('SP500.csv', index=False)
gold.to_csv('Gold.csv', index=False)

print("CSV files saved successfully.")

```
